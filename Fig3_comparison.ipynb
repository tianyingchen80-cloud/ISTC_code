## Time Lorenz L=100 ROC comparison

Compare ISTC with PCMCI and Lagged CCM by ROC curves on the three saved Lorenz L=100 experiments. All methods produce delay-specific scores, so the comparison aggregates all six `t_idx` slices into one ROC analysis per experiment.

```python
# Reload the Lorenz generators used by the saved experiments.
%run ./time_Lorenz_data_6time.ipynb
%run ./time_Lorenz_data_2time_2_3.ipynb
%run ./time_Lorenz_data_2time_3_5.ipynb

import importlib
import time_Lorenz_ROC_comparison
importlib.reload(time_Lorenz_ROC_comparison)

from time_Lorenz_ROC_comparison import (
    load_time_lorenz_l100_cases,
    compute_or_load_time_lorenz_baselines,
    plot_time_lorenz_roc_comparison,
)

# Aggregate all six TDInet time indices. With L=100, these correspond to 0, 100, 200, 300, 400, and 500 steps.
T_IDXS = range(6)
BASELINE_STRIDE = 100
CCM_EMBEDDING_DIM = 3
CCM_LIBRARY_FRACTION = 0.8
CCM_REPETITIONS = 30
FORCE_BASELINES = False

cases = load_time_lorenz_l100_cases(base_dir=".")
generators = {
    "gen_data_norm_6time": gen_data_norm_6time,
    "gen_data_norm_2time_3_5": gen_data_norm_2time_3_5,
    "gen_data_norm_2time_2_3": gen_data_norm_2time_2_3,
}

# PCMCI and Lagged CCM both return one continuous score per edge and candidate delay.
baseline_scores = compute_or_load_time_lorenz_baselines(
    cases,
    generators,
    cache_path="result/Comparison_results/time_lorenz_l100_pcmci_ccm.pkl",
    methods=("pcmci", "ccm"),
    force=FORCE_BASELINES,
    strict_methods=True,
    baseline_stride=BASELINE_STRIDE,
    ccm_embedding_dim=CCM_EMBEDDING_DIM,
    ccm_library_fraction=CCM_LIBRARY_FRACTION,
    ccm_repetitions=CCM_REPETITIONS,
)

figs = plot_time_lorenz_roc_comparison(
    cases,
    baseline_scores,
    t_idxs=T_IDXS,
    baseline_stride=BASELINE_STRIDE,
    save_path="result/figure/time_lorenz_L100_ROC.png",
    figure_width_mm=150,
    figure_height_mm=130,
)
```
