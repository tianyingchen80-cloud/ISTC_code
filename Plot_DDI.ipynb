```python
import os
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.colors as colors
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.patches import Rectangle
from matplotlib.ticker import ScalarFormatter, MaxNLocator

FIGURE_DIR = os.path.join('result', 'figure')
os.makedirs(FIGURE_DIR, exist_ok=True)
```

```python
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Times New Roman"]
plt.rcParams["font.size"] = 16
```

```python
def plot_array_1x4_Further_experiment(Arrays, ColorArrays, titles, variables, seq, threshold):

    cmap = LinearSegmentedColormap.from_list(
        'prl_blue_clean',
        ['#f7f9fb', '#d9e6f2', '#8fb3d1', '#3f78a8', '#163a5f']
    )

    fig, axs = plt.subplots(1, 4, figsize=(13.34, 3.5), constrained_layout=True)
    ax_positions = axs.ravel(order="C") 

    vmin = min(A.min() for A in Arrays)
    vmax = max(A.max() for A in Arrays)
    norm = colors.Normalize(vmin=vmin, vmax=vmax)
    ticks = [vmin, (vmin + vmax) / 2, vmax]

    shared_mappable = None
    shared_axes = []

    for k in range(4):
        A = Arrays[k]
        ax = ax_positions[k]

        pos = ax.imshow(A, cmap=cmap, norm=norm)
        if shared_mappable is None:
            shared_mappable = pos
        shared_axes.append(ax)

        ax.set_xticks(ticks=seq, labels=variables[seq])
        ax.set_yticks(ticks=seq, labels=variables[seq])
        ax.set_xlabel("Cause")
        ax.set_ylabel("Effect")
        ax.set_title(titles[k], fontsize=14)

        n = A.shape[0]
        for ii in range(n):
            for jj in range(n):
                if A[ii, jj] > threshold:
                    rect = patches.Rectangle(
                        (jj - 0.5, ii - 0.5), 1, 1,
                        linewidth=2.2,
                        edgecolor="#E68613",
                        facecolor="none",
                        linestyle="--"
                    )
                    ax.add_patch(rect)

    cbar = fig.colorbar(
        shared_mappable,
        ax=shared_axes,
        orientation="vertical",
        shrink=0.9,
        location="right",
        ticks=ticks,
        format="%.4f"
    )
    cbar.ax.tick_params(labelsize=10)

    filename = "Further_experiment_1"
    pathname = os.path.join(FIGURE_DIR, f"{filename}.png")
    plt.savefig(pathname, bbox_inches="tight", dpi=600)
    plt.show()




def plot_array_1x4_Further_experiment_2(Arrays, ColorArrays, titles, variables, seq, threshold, color_bar):

    cmap = LinearSegmentedColormap.from_list(
        'prl_purple_blue',
        ['#f8f8fb', '#f8f8fb', '#a9addd', '#646bb0', '#2b2f77']
    )

    fig, axs = plt.subplots(1, 4, figsize=(13.34, 3.5), constrained_layout=True)
    ax_positions = axs.ravel(order="C")

    vmin = min(A.min() for A in Arrays)
    vmax = max(A.max() for A in Arrays)
    vmax = max(vmax, color_bar)

    norm = colors.Normalize(vmin=vmin, vmax=vmax)

    ticks = [vmin, (vmin + vmax) / 2, vmax]

    shared_mappable = None
    shared_axes = []

    for k in range(4):
        A = Arrays[k]
        ax = ax_positions[k]

        pos = ax.imshow(A, cmap=cmap, norm=norm)
        if shared_mappable is None:
            shared_mappable = pos
        shared_axes.append(ax)

        ax.set_xticks(seq, variables[seq])
        ax.set_yticks(seq, variables[seq])
        ax.set_xlabel("Cause")
        ax.set_ylabel("Effect")
        ax.set_title(titles[k], fontsize=14)

        n = A.shape[0]
        for ii in range(n):
            for jj in range(n):
                if A[ii, jj] > threshold:
                    rect = patches.Rectangle(
                        (jj - 0.5, ii - 0.5), 1, 1,
                        linewidth=2.0,
                        edgecolor="#C65D21",
                        facecolor="none",
                        linestyle="--"
                    )
                    ax.add_patch(rect)


    
    cbar = fig.colorbar(
        shared_mappable,
        ax=shared_axes,
        orientation="vertical",
        shrink=0.9,
        location="right",
        ticks=ticks,
        format="%.4f"
    )
    cbar.ax.tick_params(labelsize=10)

    filename = "Further_experiment_2"
    pathname = os.path.join(FIGURE_DIR, f"{filename}.png")
    plt.savefig(pathname, bbox_inches="tight", dpi=600)
    plt.show()




def plot_array_1x4_Further_experiment_3(Arrays, ColorArrays, titles, variables, seq, threshold, color_bar):

    cmap = LinearSegmentedColormap.from_list(
        'prl_earth',
        ['#f8f8fb', '#f8f8fb', '#c89f7a', '#9a6448', '#5c3427']
    )

    fig, axs = plt.subplots(1, 4, figsize=(13.34, 3.5), constrained_layout=True)
    ax_positions = axs.ravel(order="C")


    vmin = min(A.min() for A in Arrays)
    vmax = max(A.max() for A in Arrays)
    vmax = max(vmax, color_bar)
    norm = colors.Normalize(vmin=vmin, vmax=vmax)
    ticks = [vmin, (vmin + vmax) / 2, vmax]

    shared_mappable = None
    shared_axes = []

    for k in range(4):
        A = Arrays[k]
        ax = ax_positions[k]

        pos = ax.imshow(A, cmap=cmap, norm=norm)
        if shared_mappable is None:
            shared_mappable = pos
        shared_axes.append(ax)

        ax.set_xticks(seq, variables[seq])
        ax.set_yticks(seq, variables[seq])
        ax.set_xlabel("Cause")
        ax.set_ylabel("Effect")
        ax.set_title(titles[k], fontsize=14)

        n = A.shape[0]
        for ii in range(n):
            for jj in range(n):
                if A[ii, jj] > threshold:
                    rect = patches.Rectangle(
                        (jj - 0.5, ii - 0.5), 1, 1,
                        linewidth=2.0,
                        edgecolor="#C65D21",
                        facecolor="none",
                        linestyle="--"
                    )
                    ax.add_patch(rect)

    cbar = fig.colorbar(
        shared_mappable,
        ax=shared_axes,
        orientation="vertical",
        shrink=0.9,
        location="right",
        ticks=ticks,
        format="%.4f"
    )
    cbar.ax.tick_params(labelsize=10)

    filename = "Further_experiment_3"
    pathname = os.path.join(FIGURE_DIR, f"{filename}.png")
    plt.savefig(pathname, bbox_inches="tight", dpi=600)
    plt.show()


def plot_bar_windowlength_Further_experiment(
    window_length_list,
    values,
    filename,
    bar_face,
    bar_edge
):
    bar_face = bar_face
    bar_edge = bar_edge
    highlight_edge = '#C65D21'

    fig, ax = plt.subplots(figsize=(4.20, 4.20), constrained_layout=True)

    fig.subplots_adjust(
        left=0.13,
        right=0.98,
        bottom=0.22,
        top=0.88
    )

    bars = ax.bar(
        window_length_list,
        values,
        width=5.0,
        color=bar_face,
        edgecolor=bar_edge,
        linewidth=1.5
    )

    max_idx = int(np.argmax(values))
    max_idx = int(np.argmax(values))
    b = bars[max_idx]

    b.set_edgecolor('none')
    b.set_linewidth(0)



    highlight_rect = Rectangle(
        (b.get_x(), 0),
        b.get_width(),
        b.get_height(),
        fill=False,
        edgecolor=highlight_edge,
        linewidth=2.2,
        linestyle='--',
        zorder=10,
        clip_on=False
    )

    ax.add_patch(highlight_rect)

    class OneDecimalScalarFormatter(ScalarFormatter):
        def _set_format(self):
            self.format = r'$\mathdefault{%.1f}$'

    formatter = OneDecimalScalarFormatter(useMathText=True)
    formatter.set_scientific(True)
    formatter.set_powerlimits((0, 0))
    formatter.set_useOffset(False)

    ax.yaxis.set_major_formatter(formatter)
    ax.yaxis.set_major_locator(MaxNLocator(nbins=5))
    ax.yaxis.get_offset_text().set_fontsize(10)

    ax.set_xlabel('Time delay', fontsize=12)
    ax.set_ylabel('LCI', fontsize=12)

    ax.set_xticks(window_length_list)
    ax.set_xlim(window_length_list[0] - 8, window_length_list[-1] + 8)

    ax.tick_params(axis='x', labelsize=10, rotation=45, width=1.2)
    ax.tick_params(axis='y', labelsize=10, width=1.2)

    for spine in ax.spines.values():
        spine.set_linewidth(1.2)

    os.makedirs(FIGURE_DIR, exist_ok=True)
    pathname = os.path.join(FIGURE_DIR, f"{filename}.png")
    plt.savefig(pathname, bbox_inches="tight", dpi=600)
    plt.show()

    print(f"figure has been saved to: {pathname}")








def plot_array_2x4_Coupled_Lorzen_1(Arrays, titles, variables, seq, threshold):
    #cmap = plt.cm.Blues
    cmap = LinearSegmentedColormap.from_list('4d7e54_ramp', ['#cfeadf', '#4d7e54'])
    fig, axs = plt.subplots(2, 4, figsize=(12, 6), constrained_layout=True)

    order = [0, 2, 3, 4, 1, 5, 6, 7]
    ax_positions = np.ravel(axs, order="C")  # (0,0),(0,1),(0,2),(0,3),(1,0),(1,1),(1,2),(1,3)

    norm12 = colors.Normalize(vmin=0, vmax=1)

    vmin = min(Arrays[i].min() for i in range(2, 8))
    vmax = max(Arrays[i].max() for i in range(2, 8))
    norm_rest = colors.Normalize(vmin=vmin, vmax=vmax)
    ticks_rest = [vmin, (vmin + vmax) / 2, vmax]

    shared_mappable = None
    shared_axes = []

    for place_idx, arr_idx in enumerate(order):
        A = Arrays[arr_idx]
        ax = ax_positions[place_idx]
        title = titles[arr_idx]

        if arr_idx in (0, 1):
            pos = ax.imshow(A, cmap=cmap, norm=norm12)
            fig.colorbar(pos, ax=ax, shrink=0.9, extend="neither", 
                         ticks=[0.0,  1.0], format="%.1f")
        else:
            pos = ax.imshow(A, cmap=cmap, norm=norm_rest)
            shared_axes.append(ax)
            if shared_mappable is None:
                shared_mappable = pos

        ax.set_xticks(ticks=seq, labels=variables[seq])
        ax.set_yticks(ticks=seq, labels=variables[seq])

        col_idx = place_idx % 4
        row_idx = place_idx // 4

        if col_idx == 0:
            ax.set_ylabel("Effect")
        else:
            ax.set_ylabel("")


        if row_idx == 1:
            ax.set_xlabel("Cause")
        else:
            ax.set_xlabel("")

        ax.set_title(title, fontsize=18, va="top")
        n = A.shape[0]
        for ii in range(n):
            for jj in range(n):
                if A[ii, jj] > threshold:
                    rect = patches.Rectangle((jj-0.5, ii-0.5), 1, 1,
                                             linewidth=3, edgecolor="orange",
                                             facecolor="none", linestyle="--")
                    ax.add_patch(rect)

    if shared_mappable is not None and shared_axes:
        cbar = fig.colorbar(shared_mappable, ax=shared_axes, orientation="vertical",
                            shrink=0.9, location="right", 
                            ticks=ticks_rest, format="%.4f")

    filename = "Coupled_Lorzen_1"
    pathname = os.path.join(FIGURE_DIR, f"{filename}.png")
    plt.savefig(pathname, bbox_inches="tight", dpi=600)
    plt.show()





def plot_array_2x4_Coupled_Lorzen_2(Arrays, titles, variables, seq, threshold):
    cmap = LinearSegmentedColormap.from_list('b53289_ramp', ['#fad7d3', '#b53289'])
    
    fig, axs = plt.subplots(2, 4, figsize=(12, 6), constrained_layout=True)

    order = [0, 2, 3, 4, 1, 5, 6, 7]
    ax_positions = np.ravel(axs, order="C")  

  
    norm12 = colors.Normalize(vmin=0, vmax=1)


    vmin = min(Arrays[i].min() for i in range(2, 8))
    vmax = max(Arrays[i].max() for i in range(2, 8))
    norm_rest = colors.Normalize(vmin=vmin, vmax=vmax)
    ticks_rest = [vmin, (vmin + vmax) / 2, vmax]

    shared_mappable = None
    shared_axes = []  

    for place_idx, arr_idx in enumerate(order):
        A = Arrays[arr_idx]
        ax = ax_positions[place_idx]
        title = titles[arr_idx]

        if arr_idx in (0, 1):
            pos = ax.imshow(A, cmap=cmap, norm=norm12)
            fig.colorbar(pos, ax=ax, shrink=0.9, extend="neither", 
                         ticks=[0.0, 1.0], format="%.1f")
        else:
            pos = ax.imshow(A, cmap=cmap, norm=norm_rest)
            shared_axes.append(ax)
            if shared_mappable is None:
                shared_mappable = pos

        ax.set_xticks(ticks=seq, labels=variables[seq])
        ax.set_yticks(ticks=seq, labels=variables[seq])

        col_idx = place_idx % 4
        row_idx = place_idx // 4

        if col_idx == 0:
            ax.set_ylabel("Effect")
        else:
            ax.set_ylabel("")


        if row_idx == 1:
            ax.set_xlabel("Cause")
        else:
            ax.set_xlabel("")

        ax.set_title(title, fontsize=18, va="top")

        n = A.shape[0]
        for ii in range(n):
            for jj in range(n):
                if A[ii, jj] > threshold:
                    rect = patches.Rectangle((jj-0.5, ii-0.5), 1, 1,
                                             linewidth=3, edgecolor="orange",
                                             facecolor="none", linestyle="--")
                    ax.add_patch(rect)

    if shared_mappable is not None and shared_axes:
        cbar = fig.colorbar(shared_mappable, ax=shared_axes, orientation="vertical",
                            shrink=0.9, location="right", 
                            ticks=ticks_rest, format="%.4f")


    filename = "Coupled_Lorzen_2"
    pathname = os.path.join(FIGURE_DIR, f"{filename}.png")
    plt.savefig(pathname, bbox_inches="tight", dpi=600)
    plt.show()


    plt.show()




def plot_array_2x6_Coupled_Lorzen_3(Arrays, titles, variables, seq, threshold):
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.colors as colors
    import matplotlib.patches as patches
    from matplotlib.colors import LinearSegmentedColormap

    cmap = LinearSegmentedColormap.from_list(
        'e64532_ramp', ['#f6e09d', '#e64532']
    )

    fig, axs = plt.subplots(2, 6, figsize=(18, 6), constrained_layout=True)

    order = [0, 1, 2, 6, 7, 8, 3, 4, 5, 9, 10, 11]
    ax_positions = np.ravel(axs, order="C")


    norm_first6 = colors.Normalize(vmin=0, vmax=1)

    vmin = min(Arrays[i].min() for i in range(6, 12))
    vmax = max(Arrays[i].max() for i in range(6, 12))
    norm_last6 = colors.Normalize(vmin=vmin, vmax=vmax)
    ticks_last6 = [vmin, (vmin + vmax) / 2, vmax]

    shared_mappable = None
    shared_axes = []

    for place_idx, arr_idx in enumerate(order):
        A = Arrays[arr_idx]
        ax = ax_positions[place_idx]
        title = titles[arr_idx]

        if arr_idx in range(6):
            pos = ax.imshow(A, cmap=cmap, norm=norm_first6)
            fig.colorbar(
                pos,
                ax=ax,
                shrink=0.9,
                extend="neither",
                ticks=[0.0, 1.0],
                format="%.1f"
            )

        else:
            pos = ax.imshow(A, cmap=cmap, norm=norm_last6)
            shared_axes.append(ax)
            if shared_mappable is None:
                shared_mappable = pos

        ax.set_xticks(ticks=seq, labels=variables[seq])
        ax.set_yticks(ticks=seq, labels=variables[seq])

   
        col_idx = place_idx % 6
        row_idx = place_idx // 6

        if col_idx == 0:
            ax.set_ylabel("Effect")
        else:
            ax.set_ylabel("")


        if row_idx == 1:
            ax.set_xlabel("Cause")
        else:
            ax.set_xlabel("")

        ax.set_title(title, fontsize=18, va="top")


        n = A.shape[0]
        for ii in range(n):
            for jj in range(n):
                if A[ii, jj] > threshold:
                    rect = patches.Rectangle(
                        (jj - 0.5, ii - 0.5),
                        1,
                        1,
                        linewidth=3,
                        edgecolor="blue",
                        facecolor="none",
                        linestyle="--"
                    )
                    ax.add_patch(rect)

    if shared_mappable is not None and shared_axes:
        cbar = fig.colorbar(
            shared_mappable,
            ax=shared_axes,
            orientation="vertical",
            shrink=0.9,
            location="right",
            ticks=ticks_last6,
            format="%.4f"
        )

    filename = "Coupled_Lorzen_3"
    pathname = os.path.join(FIGURE_DIR, f"{filename}.png")
    plt.savefig(pathname, bbox_inches="tight", dpi=600)
    plt.show()



```
